Wczytanie danych

```python
dane = pd.read_csv("reglin_1.csv")
x = dane["x"].to_numpy()
y = dane["y"].to_numpy()
dane.head()
```

Wyliczenie a i b metoda najmniejszych kwadratow

$$a=\frac{\sum_{i=1}^{n}(x_{i}-\overline{x})(y_{i}-\overline{y})}{\sum_{i=1}^{n}(x_{i}-\overline{x})^{2}} \qquad b=\overline{y}-a\overline{x}$$

```python
x_sr = x.mean()
y_sr = y.mean()

a = np.sum((x - x_sr) * (y - y_sr)) / np.sum((x - x_sr) ** 2)
b = y_sr - a * x_sr

print("a =", a)
print("b =", b)
```
